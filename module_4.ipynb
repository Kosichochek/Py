```python

import tkinter as tk
from tkinter import messagebox

class Node:
    def __init__(self, bus_num, driver, route):
        self.bus_num = bus_num
        self.driver = driver
        self.route = route
        self.next = None
        self.prev = None

class DoublyLinkedList:
    def __init__(self):
        self.head = None

    def add(self, bus_num, driver, route):
        new_node = Node(bus_num, driver, route)
        if not self.head:
            self.head = new_node
        else:
            current = self.head
            while current.next:
                current = current.next
            current.next = new_node
            new_node.prev = current

    def remove(self, bus_num):
        current = self.head
        while current:
            if current.bus_num == bus_num:
                if current.prev:
                    current.prev.next = current.next
                else:
                    self.head = current.next
                if current.next:
                    current.next.prev = current.prev
                return current
            current = current.next
        return None

    def get_all(self):
        data = []
        current = self.head
        while current:
            data.append(f"№{current.bus_num} | {current.driver} | Маршрут: {current.route}")
            current = current.next
        return data

class BusApp:
    def __init__(self, root):
        self.root = root
        self.root.title("Автобусный парк")
        self.park_list = DoublyLinkedList()
        self.route_list = DoublyLinkedList()

        # Поля ввода
        tk.Label(root, text="Номер автобуса:").grid(row=0, column=0)
        self.ent_num = tk.Entry(root)
        self.ent_num.grid(row=0, column=1)

        tk.Label(root, text="Водитель:").grid(row=1, column=0)
        self.ent_driver = tk.Entry(root)
        self.ent_driver.grid(row=1, column=1)

        tk.Label(root, text="Маршрут:").grid(row=2, column=0)
        self.ent_route = tk.Entry(root)
        self.ent_route.grid(row=2, column=1)

        # Кнопки
        tk.Button(root, text="Добавить в парк", command=self.add_to_park).grid(row=3, column=0, columnspan=2)
        tk.Button(root, text="==> На маршрут", command=self.to_route).grid(row=4, column=0)
        tk.Button(root, text="<== В парк", command=self.to_park).grid(row=4, column=1)

        # Списки
        tk.Label(root, text="В ПАРКЕ:").grid(row=5, column=0)
        self.lb_park = tk.Listbox(root, width=40)
        self.lb_park.grid(row=6, column=0)

        tk.Label(root, text="НА МАРШРУТЕ:").grid(row=5, column=1)
        self.lb_route = tk.Listbox(root, width=40)
        self.lb_route.grid(row=6, column=1)

    def update_lists(self):
        self.lb_park.delete(0, tk.END)
        for item in self.park_list.get_all():
            self.lb_park.insert(tk.END, item)
        
        self.lb_route.delete(0, tk.END)
        for item in self.route_list.get_all():
            self.lb_route.insert(tk.END, item)

    def add_to_park(self):
        num = self.ent_num.get()
        drv = self.ent_driver.get()
        rte = self.ent_route.get()
        if num and drv and rte:
            self.park_list.add(num, drv, rte)
            self.update_lists()
        else:
            messagebox.showwarning("Ошибка", "Заполните все поля")

    def to_route(self):
        num = self.ent_num.get()
        node = self.park_list.remove(num)
        if node:
            self.route_list.add(node.bus_num, node.driver, node.route)
            self.update_lists()

    def to_park(self):
        num = self.ent_num.get()
        node = self.route_list.remove(num)
        if node:
            self.park_list.add(node.bus_num, node.driver, node.route)
            self.update_lists()

if __name__ == "__main__":
    root = tk.Tk()
    app = BusApp(root)
    root.mainloop()
```

```python

import random
import time

class Queue:
    def __init__(self):
        self.items = []

    def enqueue(self, item):
        self.items.append(item)

    def dequeue(self):
        if not self.is_empty():
            return self.items.pop(0)
        return None

    def is_empty(self):
        return len(self.items) == 0

    def size(self):
        return len(self.items)

# Параметры моделирования
total_work_time = int(input("Введите общее время работы кассы (мин): "))
service_per_client = int(input("Введите время обслуживания одного клиента (мин): "))
check_interval = int(input("Введите интервал проверки очереди (мин): "))

shop_queue = Queue()
current_time = 0
busy_until = 0
# Время прибытия первого клиента (нормальное распределение: среднее 5 мин, откл. 2 мин)
next_arrival = max(1, int(random.gauss(5, 2)))

print("\n--- Моделирование работы очереди ---")
print("Время (мин) | Длина очереди")
print("---------------------------")

while current_time <= total_work_time:
    # Прибытие нового клиента
    if current_time == next_arrival:
        shop_queue.enqueue("Клиент")
        # Рассчитываем время до следующего клиента
        arrival_delta = max(1, int(random.gauss(5, 2)))
        next_arrival += arrival_delta

    # Обслуживание клиента
    if current_time >= busy_until and not shop_queue.is_empty():
        shop_queue.dequeue()
        busy_until = current_time + service_per_client

    # Вывод состояния через интервал
    if current_time % check_interval == 0:
        print(f"{current_time:11} | {shop_queue.size():13}")

    current_time += 1

print("---------------------------")
print("Касса закрыта.")
```

```python

class Node:
    def __init__(self, name, position, emp_id, salary):
        self.data = {
            'name': name,
            'position': position,
            'id': emp_id,
            'salary': salary
        }
        self.next = None

class SinglyLinkedList:
    def __init__(self):
        self.head = None

    def append(self, name, position, emp_id, salary):
        new_node = Node(name, position, emp_id, salary)
        if not self.head:
            self.head = new_node
            return
        last = self.head
        while last.next:
            last = last.next
        last.next = new_node

    def get_length(self):
        count = 0
        current = self.head
        while current:
            count += 1
            current = current.next
        return count

    def search(self, name):
        current = self.head
        index = 1
        while current:
            if current.data['name'].lower() == name.lower():
                return index
            current = current.next
            index += 1
        return -1

# Основная часть программы
emp_list = SinglyLinkedList()

print("--- Создание списка сотрудников ---")
while True:
    name = input("Введите имя (или 'стоп' для завершения): ")
    if name.lower() == 'стоп':
        break
    pos = input("Введите должность: ")
    e_id = int(input("Введите ID: "))
    sal = int(input("Введите оклад: "))
    emp_list.append(name, pos, e_id, sal)

print(f"\nТекущая длина списка: {emp_list.get_length()}")

search_name = input("\nВведите имя сотрудника для поиска: ")
result = emp_list.search(search_name)

if result != -1:
    print(f"Сотрудник найден под номером: {result}")
else:
    print("Данные не найдены.")
```

```python

class Stack:
    def __init__(self):
        self.items = []

    def push(self, item):
        self.items.append(item)

    def pop(self):
        if not self.is_empty():
            return self.items.pop()
        return None

    def is_empty(self):
        return len(self.items) == 0

size = int(input("Введите размер массива: "))
array = []
for i in range(size):
    array.append(input(f"Введите элемент {i + 1}: "))

print("Исходный массив:", array)

while True:
    invert_choice = input("Инвертировать исходный массив? (Да/Нет): ")
    if invert_choice.strip().lower() == "да":
        stack = Stack()
        for element in array:
            stack.push(element)
        
        inverted_array = []
        while not stack.is_empty():
            inverted_array.append(stack.pop())
        print("Инвертированный массив:", inverted_array)

    change_choice = input("Изменить исходный массив (добавить/удалить)? (Да/Нет): ")
    if change_choice.strip().lower() == "да":
        action = input("Желаете 'добавить' или 'удалить' элемент? ").strip().lower()
        new_array = list(array)
        
        if action == "добавить":
            new_item = input("Введите элемент для добавления: ")
            new_array.append(new_item)
        elif action == "удалить":
            item_to_remove = input("Введите элемент дл я удаления: ")
            if item_to_remove in new_array:
                new_array.remove(item_to_remove)
            else:
                print("Элемент не найден в массиве.")
        
        array = new_array
        print("Новый массив:", array)

    exit_choice = input("Выйти из приложения? (Да/Нет): ")
    if exit_choice.strip().lower() == "да":
        break
```
